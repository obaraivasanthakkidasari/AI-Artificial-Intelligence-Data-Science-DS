# left Tail Test

Question 

A machine is producing perfume bottles with the long-term average of 150 ml and the standard deviation of 2 ml. 
Four bottles were picked and the average volume was found to be 149 ml. 
Has the average volume minimum of 150 ml? Test with 95% confidence level.

Solution
- Null Hypothesis (H₀): Mean volume ≥ 150 ml
- Alternative Hypothesis (H₁): Mean volume < 150 ml
- Alpha (α) = 1 − 0.95 = 0.05
- Test: Left Tail Z-Test
- Z calculated = (Observed mean − Population mean) / Standard error
- Formula: z = (x̄ − μ) / (σ / √n)

```python
z_cal = (149-150)/(2/np.sqrt(4))
z_cal
```

```python
#Left tail test -----> stats.norm.cdf(z_cal) -----> Cummulative Distribution Function
stats.norm.cdf(-1.0)
```

Inferences
- P-value = 0.1587
- Alpha = 0.05
- P-value > Alpha (0.1587 > 0.05)
- Do not reject H₀.

## Right Tail Test

Question
The mean response time for a random sample of 30 pizza deliveries is 32 minutes. The population standard deviation is believed to be 10 minutes. The pizza delivery services director wants to perform a hypothesis test, with a 0.05 level of significance, to determine whether the service goal of 30 minutes or less is being achieved.
Solution
- Null Hypothesis (H₀): Mean time ≤ 30 minutes
- Alternative Hypothesis (H₁): Mean time > 30 minutes
- Alpha (α) = 0.05
- Test: Right Tail Z-Test — Right Tail Rejection
- Z calculated = (Observed mean − Population mean) / Standard error
- Formula: z = (x̄ − μ) / (σ / √n)
Python Code
Step 1: Calculate Z-score

```python
Z_cal = (32-30)/(10/np.sqrt(30))
print(Z_cal)
```

```python
#Right  tail test -----> 1-norm.cdf(z_cal) -----> Cummulative Distribution Function
1- stats.norm.cdf(1.09)
```

```python
#Right  tail test -----> norm.cdf(z_cal) -----> Survival function.
stats.norm.sf(1.09)
```

#Inferences (Right Tail Test)
- P-value is high  Do not reject the null hypothesis.
- Do not reject H₀.
- There is insufficient statistical evidence to infer that pizza delivery services are not meeting the response 
- goal of 30 minutes.

# Two Tail Test
Question
Assume that a sample of 30 milk cartons provides a sample mean of 505 ml.
- The population standard deviation is believed to be 10 ml.
- Perform a hypothesis test at the 0.03 level of significance, with a hypothesized population mean of 500 ml, to help determine whether the filling process should continue operating or be stopped and corrected.
Solution
- Null Hypothesis (H₀): Mean = 500 ml
- Alternative Hypothesis (H₁): Mean ≠ 500 ml
- Test: Two Tail Z-Test
- Alpha (α) = 0.03
- Z calculated = (Observed mean − Population mean) / Standard error
- Formula: z = (x̄ − μ) / (σ / √n)

```python
Z_cal = (505-500)/(10/np.sqrt(30))
print(Z_cal)
```

```python
# Two Tail -----> (norms.sf(Z_cal))*2
pvalue = 2*stats.norm.sf(Z_cal)
pvalue
```

Inferences
- P-value = 0.00617
- Alpha = 0.03
- P-value < Alpha (0.00617 < 0.03)
- Reject H₀.

Conclusion
At the 3% significance level (97% confidence level), there is sufficient statistical evidence to conclude that the 
average milk carton volume is significantly different from 500 ml.

Therefore, reject the null hypothesis (H₀). The filling process should be investigated and 
corrected if necessary because the average filling volume appears to differ from the target of 500 ml.
